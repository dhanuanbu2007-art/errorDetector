```python
import gradio as gr
import language_tool_python


# ============================================
# LOAD GRAMMAR CHECKER
# ============================================

print("Starting Grammar Checker...")

tool = language_tool_python.LanguageTool("en-US")

print("Grammar Checker Ready!")


# ============================================
# GRAMMAR ANALYSIS
# ============================================

def check_grammar(text):

    if not text or not text.strip():
        return (
            "Please enter some text.",
            "No errors found.",
            "No corrections available.",
            "0 errors"
        )

    text = text.strip()

    # Find grammar/spelling errors
    matches = tool.check(text)

    # Automatically correct the text
    corrected_text = tool.correct(text)

    # Prepare error details
    error_details = []

    for i, match in enumerate(matches, start=1):

        original = text[
            match.offset:
            match.offset + match.errorLength
        ]

        suggestions = match.replacements[:3]

        suggestion_text = ", ".join(suggestions)

        error_details.append(
            f"{i}. ❌ {original}\n"
            f"   Problem: {match.message}\n"
            f"   Suggestion: {suggestion_text}"
        )

    if error_details:
        details = "\n\n".join(error_details)
    else:
        details = "✅ No grammar or spelling errors detected."

    error_count = f"{len(matches)} error(s) detected"

    return (
        text,
        details,
        corrected_text,
        error_count
    )


# ============================================
# GRADIO INTERFACE
# ============================================

with gr.Blocks(
    title="Grammar Error Detector"
) as app:

    gr.Markdown(
        """
        # ✍️ Grammar Error Detector

        Enter a sentence or paragraph to detect
        grammar and spelling mistakes.

        The application will show the detected errors,
        suggestions, and corrected text.
        """
    )

    # Input
    text_input = gr.Textbox(
        label="📝 Enter Your Text",
        placeholder="Example: She go to college every day.",
        lines=8
    )

    # Button
    check_button = gr.Button(
        "🔍 Check Grammar",
        variant="primary"
    )

    gr.Markdown("## 📊 Grammar Analysis")

    # Original text
    original_output = gr.Textbox(
        label="📄 Original Text",
        lines=5
    )

    # Errors
    error_output = gr.Textbox(
        label="❌ Detected Errors & Suggestions",
        lines=10
    )

    # Corrected text
    corrected_output = gr.Textbox(
        label="✅ Corrected Text",
        lines=6
    )

    # Error count
    count_output = gr.Textbox(
        label="📊 Error Count"
    )

    # Connect button
    check_button.click(
        fn=check_grammar,
        inputs=text_input,
        outputs=[
            original_output,
            error_output,
            corrected_output,
            count_output
        ]
    )


# ============================================
# LAUNCH APPLICATION
# ============================================

app.launch(share=True)
```
